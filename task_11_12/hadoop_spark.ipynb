```python
from pyspark.sql.functions import col, avg, count, round
import urllib.request
import subprocess

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
local_path = "/tmp/titanic.csv"
urllib.request.urlretrieve(url, local_path)
print("Файл скачан")

hdfs_path = "hdfs:///user/zeppelin/titanic.csv"
subprocess.run(["hdfs", "dfs", "-put", "-f", local_path, hdfs_path], check=True)
print("Файл загружен в HDFS:", hdfs_path)

df = spark.read.csv(hdfs_path, header=True, inferSchema=True)
print("\nИсходные данные (первые 10 строк)")
df.show(10)

transformed_df = df.filter(col("Age").isNotNull()) \
    .groupBy("Pclass") \
    .agg(
        count("PassengerId").alias("Total_Passengers"),
        round(avg("Age"), 1).alias("Average_Age"),
        round(avg("Fare"), 2).alias("Average_Fare")
    ).orderBy("Pclass")

print("\nСтатистика по классам кают")
transformed_df.show()

output_path = "hdfs:///user/zeppelin/titanic_result.parquet"
transformed_df.write.mode("overwrite").parquet(output_path)
print("Результат сохранён в HDFS:", output_path)

result_check = spark.read.parquet(output_path)
print("\nПроверка записанного файла")
result_check.show()
```
