```scala
%python
from kafka import KafkaProducer
import json
import random
import time
from datetime import datetime

KAFKA_BROKERS = ['rc1b-j135r7jtv9gs67b2.mdb.yandexcloud.net:9091']
KAFKA_USER = 'mkf-user'
KAFKA_PASSWORD = 'mkf-user'
TOPIC = 'sensors'

producer = KafkaProducer(
    bootstrap_servers=KAFKA_BROKERS,
    security_protocol='SASL_SSL',
    sasl_mechanism='SCRAM-SHA-512',
    sasl_plain_username=KAFKA_USER,
    sasl_plain_password=KAFKA_PASSWORD,
    ssl_cafile='CA.pem',
    value_serializer=lambda v: json.dumps(v).encode('utf-8'),
    key_serializer=lambda k: k.encode('utf-8')
)

print("Генерацию и отправку данных:")

for i in range(1, 6):
    device = f"device_abc_{i}"
    data = {
        "device_id": device,
        "datetime": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "latitude": round(random.uniform(55.0, 56.0), 6),
        "longitude": round(random.uniform(37.0, 38.0), 6),
        "altitude": round(random.uniform(100.0, 200.0), 1),
        "speed": round(random.uniform(20.0, 100.0), 1),
        "battery_voltage": round(random.uniform(12.0, 14.5), 1),
        "cabin_temperature": random.randint(18, 25),
        "fuel_level": random.randint(10, 80)
    }

    producer.send(TOPIC, key=device, value=data)
    print(f"Отправлено: {data}")
    time.sleep(1)

producer.flush()
print("Все данные улетели в Kafka (и через трансфер в MongoDB)")
```
